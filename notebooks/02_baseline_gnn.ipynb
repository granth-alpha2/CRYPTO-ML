# 02 — Baseline GNN Training & Evaluation
### Plaintext GCN / GIN on Elliptic Dataset

This notebook trains a plaintext GNN and evaluates it — establishing
the **Stage 1 baseline** before quantization and encryption.

Steps:
1. Load processed PyG graph
2. Build and train GCN model
3. Plot training curves (loss, F1)
4. Confusion matrix and ROC curve
5. Compare with quantized model (Stage 2)

**Prerequisites:** Run `python src/preprocess.py` first.

```python
import sys, os
sys.path.insert(0, '..')

import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    confusion_matrix, classification_report,
    roc_curve, auc, f1_score
)

from src.utils import load_config, set_seed, print_metrics_table
from src.preprocess import load_processed
from src.model import build_model
from src.quantize import quantize_model, swap_activations

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 120
set_seed(42)
print('Imports OK')
```

## 1. Load Config and Data

```python
cfg  = load_config('../configs/config.yaml')
data, scaler = load_processed('../' + cfg['paths']['processed_data'])
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Graph: {data.num_nodes:,} nodes, {data.num_edges:,} edges, {data.num_features} features')
print(f'Train/Val/Test: {data.train_mask.sum()}/{data.val_mask.sum()}/{data.test_mask.sum()}')
print(f'Device: {device}')
```

## 2. Build Model

```python
model = build_model(cfg).to(device)
n_params = sum(p.numel() for p in model.parameters())
print(model)
print(f'\nTotal parameters: {n_params:,}')
```

## 3. Train (or Load Pre-trained)

```python
MODEL_PATH = '../' + cfg['paths']['model_save']

if os.path.exists(MODEL_PATH):
    print(f'Loading pre-trained model from {MODEL_PATH}')
    model.load_state_dict(torch.load(MODEL_PATH, map_location=device))
else:
    print('No saved model found. Training now...')
    from src.train import train
    cfg_local = load_config('../configs/config.yaml')
    cfg_local['paths'] = {k: '../'+v for k,v in cfg_local['paths'].items()}
    model = train(cfg_local).to(device)
```

## 4. Plot Training Curves

```python
metrics_path = '../results/metrics.csv'
if os.path.exists(metrics_path):
    df_metrics = pd.read_csv(metrics_path)

    fig, axes = plt.subplots(1, 2, figsize=(14, 4))

    # Loss
    axes[0].plot(df_metrics['epoch'], df_metrics['loss'], color='steelblue', linewidth=2)
    axes[0].set_title('Training Loss', fontweight='bold')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Cross-Entropy Loss')

    # F1
    axes[1].plot(df_metrics['epoch'], df_metrics['train_f1'],
                 label='Train F1', color='steelblue', linewidth=2)
    axes[1].plot(df_metrics['epoch'], df_metrics['val_f1'],
                 label='Val F1', color='darkorange', linewidth=2, linestyle='--')
    axes[1].set_title('F1 Score (Macro)', fontweight='bold')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Macro F1')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig('../results/plots/02_training_curves.png', dpi=150)
    plt.show()
else:
    print(f'No metrics file found at {metrics_path}. Train the model first.')
```

## 5. Evaluate on Test Set — Stage 1 (Plaintext GCN)

```python
import torch.nn.functional as F

model.eval()
with torch.no_grad():
    out   = model(data.x.to(device), data.edge_index.to(device)).cpu()
    probs = F.softmax(out, dim=1).numpy()
    preds = out.argmax(dim=1).numpy()

mask   = data.test_mask.numpy()
y_true = data.y.numpy()[mask]
y_pred = preds[mask]
y_prob = probs[mask, 1]

print('Stage 1 — Plaintext GCN')
print(classification_report(y_true, y_pred, target_names=['Licit','Illicit']))
```

## 6. Confusion Matrix

```python
cm = confusion_matrix(y_true, y_pred)
fig, ax = plt.subplots(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Licit','Illicit'],
            yticklabels=['Licit','Illicit'], ax=ax)
ax.set_title('Confusion Matrix — Stage 1 (Plaintext GCN)', fontweight='bold')
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
plt.tight_layout()
plt.savefig('../results/plots/02_confusion_matrix.png', dpi=150)
plt.show()
```

## 7. ROC Curve

```python
fpr, tpr, _ = roc_curve(y_true, y_prob)
roc_auc = auc(fpr, tpr)

fig, ax = plt.subplots(figsize=(7,6))
ax.plot(fpr, tpr, color='steelblue', lw=2, label=f'GCN (AUC = {roc_auc:.3f})')
ax.plot([0,1],[0,1], color='gray', linestyle='--', lw=1, label='Random')
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curve — Stage 1 Plaintext GCN', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.savefig('../results/plots/02_roc_curve.png', dpi=150)
plt.show()
```

## 8. Stage 2 — Quantized Model Comparison

```python
# Quantize and swap activations
model_q = quantize_model(model.cpu(), bits=cfg['quantize']['bits'])
model_q = swap_activations(model_q)

model_q.eval()
with torch.no_grad():
    out_q   = model_q(data.x, data.edge_index)
    preds_q = out_q.argmax(dim=1).numpy()

y_pred_q = preds_q[mask]
f1_base  = f1_score(y_true, y_pred,   average='macro', zero_division=0)
f1_quant = f1_score(y_true, y_pred_q, average='macro', zero_division=0)

print_metrics_table({
    'Stage 1 F1 (plaintext)': round(f1_base, 4),
    'Stage 2 F1 (quantized)': round(f1_quant, 4),
    'Delta (accuracy lost)' : round(f1_base - f1_quant, 4),
}, title='Quantization Impact')
```
