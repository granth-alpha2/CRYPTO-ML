# 01 — Exploratory Data Analysis
### Elliptic Bitcoin Transaction Dataset

This notebook explores the raw Elliptic dataset before any preprocessing.
Goals:
- Understand the scale and structure of the graph
- Visualise class distribution (licit / illicit / unknown)
- Understand feature distributions
- Identify temporal patterns across time steps

**Run `python src/preprocess.py` before this notebook.**

```python
import os
import sys
sys.path.insert(0, '..')   # project root

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120
print('Imports OK')
```

## 1. Load Raw CSVs

```python
RAW = '../data/raw/'

feat_cols = ['txId'] + [f'f{i}' for i in range(1, 167)]
df_feats  = pd.read_csv(RAW + 'elliptic_txs_features.csv', header=None, names=feat_cols)
df_edges  = pd.read_csv(RAW + 'elliptic_txs_edgelist.csv', header=None, names=['src','dst'])
df_cls    = pd.read_csv(RAW + 'elliptic_txs_classes.csv')

print(f'Nodes    : {len(df_feats):,}')
print(f'Edges    : {len(df_edges):,}')
print(f'Classes  : {df_cls["class"].value_counts().to_dict()}')
```

## 2. Class Distribution

```python
label_map = {'1': 'Illicit', '2': 'Licit', 'unknown': 'Unknown'}
df_cls['label'] = df_cls['class'].astype(str).map(label_map)
counts = df_cls['label'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Bar chart
colors = ['#e74c3c', '#2ecc71', '#95a5a6']
axes[0].bar(counts.index, counts.values, color=colors)
axes[0].set_title('Transaction Class Distribution', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Count')
for i, (idx, val) in enumerate(counts.items()):
    axes[0].text(i, val + 500, f'{val:,}', ha='center', fontweight='bold')

# Pie chart (labeled only)
labeled = counts.drop('Unknown', errors='ignore')
axes[1].pie(labeled, labels=labeled.index, autopct='%1.1f%%',
            colors=['#e74c3c','#2ecc71'], startangle=90)
axes[1].set_title('Labeled Transactions Only', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig('../results/plots/01_class_distribution.png', dpi=150)
plt.show()
print('Class distribution plot saved.')
```

## 3. Temporal Analysis (Time Steps)

```python
# Feature f1 is the time step in the Elliptic dataset
df = df_feats.merge(df_cls, on='txId', how='left')
df['label'] = df['class'].astype(str).map(label_map)
df['timestep'] = df['f1'].astype(int)

pivot = df.groupby(['timestep','label']).size().unstack(fill_value=0)

fig, ax = plt.subplots(figsize=(14, 5))
pivot[['Licit','Illicit','Unknown']].plot(kind='bar', stacked=True, ax=ax,
    color=['#2ecc71','#e74c3c','#95a5a6'], width=0.85)
ax.set_title('Transactions per Time Step (1-49)', fontsize=13, fontweight='bold')
ax.set_xlabel('Time Step')
ax.set_ylabel('Number of Transactions')
ax.legend(loc='upper right')
plt.tight_layout()
plt.savefig('../results/plots/01_temporal.png', dpi=150)
plt.show()
```

## 4. Feature Distribution (Sample Features)

```python
sample_feats = ['f2','f3','f4','f5','f6']
df_labeled = df[df['label'].isin(['Licit','Illicit'])]

fig, axes = plt.subplots(1, 5, figsize=(18, 4))
for ax, feat in zip(axes, sample_feats):
    for label, color in [('Licit','#2ecc71'),('Illicit','#e74c3c')]:
        subset = df_labeled[df_labeled['label']==label][feat]
        ax.hist(subset, bins=40, alpha=0.6, label=label, color=color, density=True)
    ax.set_title(feat)
    ax.set_xlabel('Value')
    ax.legend(fontsize=7)

fig.suptitle('Feature Distributions: Licit vs Illicit', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('../results/plots/01_feature_dists.png', dpi=150)
plt.show()
```

## 5. Graph Statistics

```python
# Degree distribution
from collections import Counter

degree = Counter(df_edges['src'].tolist() + df_edges['dst'].tolist())
deg_vals = list(degree.values())

fig, ax = plt.subplots(figsize=(10,4))
ax.hist(deg_vals, bins=50, log=True, color='steelblue', edgecolor='white')
ax.set_title('Node Degree Distribution (log scale)', fontsize=13, fontweight='bold')
ax.set_xlabel('Degree (number of edges)')
ax.set_ylabel('Count (log)')
plt.tight_layout()
plt.savefig('../results/plots/01_degree_distribution.png', dpi=150)
plt.show()

print(f'Mean degree : {np.mean(deg_vals):.2f}')
print(f'Max degree  : {max(deg_vals):,}')
print(f'Avg edges/node: {len(df_edges)/len(df_feats):.2f}')
```
